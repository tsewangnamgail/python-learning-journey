**Data Cleaning**

Goal: Turn messy, inconsistent, or incomplete data into a clean dataset that can be used for analysis and machine learning.

```python
#Create a Messy Dataset
import pandas as pd

data = {
    "Name": ["Tenzin", "Rahul", "Priya", "Arun", "Tenzin", None],
    "Age": [20, 21, "20", 22, 20, 21],
    "Department": ["CSE", "cse", "CSE", "IT", "CSE", "IT"],
    "Marks": [85, None, 92, "74", 85, 88],
    "City": ["Leh", "Delhi", "Leh", None, "Leh", "Delhi"]
}

df = pd.DataFrame(data)

print(df)
```

**isna()**

isna() detects missing values.

```python
print(df.isna())

#Count missing values
print(df.isna().sum())
```

**notna()**

notna() is the opposite of isna().

```python
#This selects rows where Name is not missing.
print(df[df["Name"].notna()])
```

**dropna()**

dropna() removes rows containing missing values.

```python
print(df.dropna())

#dropna() does not normally modify the original DataFrame.
clean_df = df.dropna()
print(clean_df)

#Drop rows where a specific column is missing
df.dropna(subset=["Marks"])
```

**fillna()**

Instead of deleting missing values, we can fill them.

```python
# Let's clean the column immediately so it becomes true numbers
# 'errors="coerce"' turns any unreadable text/strings into NaN safely
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# 1. Fill missing numerical values with a static 0
df["Marks_with_0"] = df["Marks"].fillna(0)

# 2. Fill text missing values with "Unknown"
df["City_with_Unknown"] = df["City"].fillna("Unknown")

# 3. Fill numerical data with the Mean (Average) -> This will NOT crash anymore
df["Marks_with_Mean"] = df["Marks"].fillna(df["Marks"].mean())

# 4. Fill numerical data with the Median (Best for handling outliers)
df["Marks_with_Median"] = df["Marks"].fillna(df["Marks"].median())

# 5. Fill text missing values with the Mode (Most frequent item)
# Fixed: added .iloc[0] to extract the raw text value from the mode series safely
df["City_with_Mode"] = df["City"].fillna(df["City"].mode().iloc[0] if not df["City"].mode().empty else "Unknown")

# Show the cleaned data
print(df.head())
```

**duplicated()**

duplicated() identifies duplicate rows.

```python
print(df.duplicated())

#Count duplicates
print(df.duplicated().sum())
```

**drop_duplicates()**

drop_duplicates() removes duplicate rows.

```python
df = df.drop_duplicates()
print(df)


#Checking Duplicates in a Specific Column
df["Name"].duplicated()
df["Name"].duplicated().sum()
df.drop_duplicates(subset=["Name"])
```

**astype()**

astype() is used to change a column's data type.

```python
print(df["Age"].dtype)

#even though Age should be numeric.
df["Age"] = df["Age"].astype(int)
print(df["Age"].dtype)

#Convert Marks
df["Marks"] = df["Marks"].astype(int)
print(df["Marks"].dtype)

#convert to float
df["Marks"] = df["Marks"].astype(float)
print(df["Marks"].dtype)

#convert to string
df["Name"] = df["Name"].astype(str)
print(df["Name"].dtype)
```

**Numerical Columns Stored as Strings**

```python
#  if marks are in string format
df["Marks"] = pd.to_numeric(df["Marks"])
```

**Handling invalid numerical values**


```python
df["Marks"] = pd.to_numeric(df["Marks"],errors="coerce")
#The invalid value becomes NaN.
```

**rename()**

rename() changes column names or index labels.

```python
df = df.rename(columns={
    "Name": "Student_Name",
    "Department": "Dept",
    "Marks": "Score"
})

#Rename one column
df = df.rename(columns={
    "Department": "Dept"
})
```

**replace()**

replace() replaces specific values.

```python
df["Department"] = df["Department"].replace({
    "cse": "CSE",
    "Cse": "CSE"
})


#Replace multiple values
df["Department"] = df["Department"].replace({
    "cse": "CSE",
    "Cse": "CSE",
    "IT ": "IT"
})


#remove extra spaces:
df["Department"] = df["Department"].str.strip()
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
)
```
