**Loading & Inspecting Data**

Learn how to load real datasets and quickly understand their structure, data types, distributions, and categorical values.

```python
import pandas as pd
```

**dataframe define**

```python
# 1. Define the raw data inside a dictionary
data = {
    'Name': [
        'Rob', 'Michael', 'Mohan', 'Ismail', 'Kory', 'Gautam', 'David', 
        'Andrea', 'Brad', 'Angelina', 'Donald', 'Tom', 'Arnold', 'Jared', 
        'Stark', 'Ranbir', 'Dipika', 'Priyanka', 'Nick', 'Alia', 'Sid', 'Abdul'
    ],
    'Age': [
        27, 29, 29, 28, 42, 39, 41, 38, 36, 35, 37, 26, 27, 28, 29, 32, 40, 41, 43, 39, 41, 39
    ],
    'Income($)': [
        70000, 90000, 61000, 60000, 150000, 155000, 160000, 162000, 156000, 
        130000, 137000, 45000, 48000, 51000, 49500, 53000, 65000, 63000, 64000, 80000, 82000, 58000
    ]
}

# 2. Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# 3.  Save the clean data directly into your specific folder path
df.to_csv("/Users/TSEGAIL/Desktop/Python/Pandas/income.csv", index=False)

print("The clean dataset has been successfully stored in your Pandas folder!")
```
