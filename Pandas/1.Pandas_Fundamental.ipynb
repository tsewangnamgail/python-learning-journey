```python
import pandas as pd  #pd is the standard alias used for Pandas.
```

**Pandas Series**

A Series is a one-dimensional labeled array.

```python
marks = pd.Series([85, 90, 78, 92])
print(marks)

#Creating Series from a List
marks = pd.Series([85, 90, 78, 92])
marks = pd.Series(
    [85, 90, 78],
    index=["Math", "Python", "DBMS"]
)
print(marks)
print(marks["Python"])

#Creating Series from a Dictionary
marks = {
    "Math": 85,
    "Python": 90,
    "DBMS": 78
}
s = pd.Series(marks)
print(s)
```

**Basic Series Information**

```python
print(s.shape)
print(s.size)
print(s.index)
print(s.values)
print(s.dtype)
```

**Pandas DataFrame**

A DataFrame is a two-dimensional labeled table.

```python
data = {
    "Name": ["Tenzin", "Rahul", "Priya"],
    "Age": [20, 21, 20],
    "Marks": [85, 92, 88]
}
df = pd.DataFrame(data)
print(df)

#Creating DataFrame from Dictionary
data = {
    "Name": ["A", "B", "C"],
    "Age": [20, 21, 22],
    "Score": [80, 90, 85]
}

df = pd.DataFrame(data)
print(df)
```

**Basic DataFrame Information**

These commands are extremely important for EDA and ML.


```python
#df.shape
print(df.shape)

#df.columns
print(df.columns)

#df.index
print(df.index)

#df.dtypes
print(df.dtypes)

#df.info()
print(df.info())

#df.describe()   ->   Provides statistical information about numerical columns.
print(df.describe())
```

**Student Dataset Practice**


```python
#creating a dataset
students = {
    "Name": ["Tenzin", "Rahul", "Priya", "Arun", "Meena"],
    "Age": [20, 21, 20, 22, 21],
    "Department": ["CSE", "ECE", "CSE", "IT", "CSE"],
    "Marks": [85, 78, 92, 74, 88]
}
df_students = pd.DataFrame(students)
print(df_students)


print(df_students.shape)
print(df_students.columns)
print(df_students.index)
print(df_students.dtypes)
print(df_students.info())
print(df_students.describe())
```
