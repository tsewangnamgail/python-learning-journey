**Selecting & Filtering Data**

Goal: Learn how to extract exactly the rows and columns you need from a Pandas DataFrame.

```python
import pandas as pd

students = {
    "Name": ["Tenzin", "Rahul", "Priya", "Arun", "Meena",
             "John", "David", "Anu"],
    "Age": [20, 21, 20, 22, 21, 20, 22, 21],
    "Department": ["CSE", "ECE", "CSE", "IT", "CSE",
                   "ECE", "IT", "CSE"],
    "Marks": [85, 78, 92, 74, 88, 81, 69, 95]
}

df = pd.DataFrame(students)

print(df)
```

**Selecting One Column**

```python
print(df["Name"])
print(df["Marks"])
```

**Selecting Multiple Columns**


```python
df[["Name", "Department", "Marks"]]
```

**.loc[]**

.loc[] is used for label-based selection.

```python
print(df.loc[2, "Name"])  #[row_label, column_label]

#Select an entire row
print(df.loc[2])  #print entire row 2

#Select multiple rows
print(df.loc[1:4])

#select specific row
print(df.loc[[1, 3, 5]])

#.loc[] — Selecting Rows and Columns
print(df.loc[1:4, ["Name", "Marks"]])
```

**.iloc[]**

.iloc[] means integer-location based indexing.

```python
print(df.iloc[2, 0])

#Select a row
print(df.iloc[2])

#Select multiple rows
print(df.iloc[1:5])

#Select a specific column by position
print(df.iloc[:, 0])
```

**Boolean Filtering**

Boolean filtering means selecting rows based on a condition.


```python
print(df["Marks"] > 80)

#Now use it to filter the DataFrame:
df[df["Marks"] > 80]

#Other Comparison Operators
print(df[df["Marks"] < 80])
print(df[df["Marks"] >= 80])
print(df[df["Marks"] == 85])
print(df[df["Marks"] != 85])
```

**Filtering Text**

You can also filter strings.

```python
print(df[df["Department"] == "CSE"])
print(df[df["Department"] != "CSE"])
```

**Multiple Conditions with &**

```python
df[(df["Department"] == "CSE") & (df["Marks"] > 85)]
```

**& vs and**


```python
#Python and
#Used for normal single Boolean expressions.
#age > 20 and marks > 80

#Pandas &
#Used for element-wise conditions on Series.
(df["Age"] > 20) & (df["Marks"] > 80)
```

**Multiple Conditions with |**

| means OR.

```python
df[
    (df["Department"] == "CSE") |
    (df["Department"] == "ECE")
]
```

**| vs or**


```python
#In Pandas:
#(condition1) | (condition2)

#In python:
#condition1 or condition2
```

**~ — Negation**

~ means NOT.

```python
df[~(df["Department"] == "CSE")]
```

**.isin()**

.isin() is useful when checking whether a value belongs to a list of values.

```python
df[(df["Department"] == "CSE") | (df["Department"] == "IT")]


#.isin() with numbers
print(df[df["Age"].isin([20, 22])])

#Negating .isin()
df[~df["Department"].isin(["CSE", "IT"])]
```

**.between()**

.between() checks whether values fall within a range.

```python
print(df[df["Marks"].between(80, 90)])
```

**Selecting Rows and Columns Simultaneously**

```python
df.loc[
    (df["Department"] == "CSE") & (df["Marks"] > 85),
    ["Name", "Marks"]
]
```

```python
#Combining Everything
result = df.loc[
    (df["Department"].isin(["CSE", "ECE"])) &
    (df["Age"].between(20, 21)) &
    (df["Marks"] > 80),
    ["Name", "Age", "Marks"]
]

print(result)
```
