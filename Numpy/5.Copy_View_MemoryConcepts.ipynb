**Copy, View, and Memory Concepts**

Goal: Understand how NumPy stores, references, and modifies data, especially when working with large datasets, images, and machine learning arrays.

**1. Views vs Copies**

NumPy arrays can share the same underlying memory. This creates two important concepts:

View: A new way to access the same data.

Copy: A completely independent duplicate of the data.

```python
#A. What Is a View?
#A view does not create a separate copy of the data. It references the original array’s memory.
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
view = arr[1:4]
print(view)

#modifying
view[0] = 999
print("View:", view)
print("Original:", arr)
```

```python
#B. What Is a Copy?
#A copy creates a separate array with its own memory.
arr = np.array([10, 20, 30, 40, 50])
copy_arr = arr.copy()
copy_arr[0] = 999
print("Copy:", copy_arr)
print("Original:", arr)
```

```python
#Checking Whether Arrays Share Memory
#np.shares_memory()
#Checks whether two arrays share memory.
arr = np.array([10, 20, 30, 40])
view = arr[1:3]
copy_arr = arr.copy()
print(np.shares_memory(arr, view))
print(np.shares_memory(arr, copy_arr))
#This function checks whether arrays may share memory. It is a faster,conservative check and may sometimes return True even when the memory overlap is uncertain.
```

**Views Created by Different Operations**

Not every NumPy operation behaves exactly the same way.

```python
#Basic Slicing
#Usually creates a view:
arr = np.array([10, 20, 30, 40])
view = arr[1:3]
view[0] = 999
print(arr)


#Fancy Indexing
#Fancy indexing generally creates a copy:
arr = np.array([10, 20, 30, 40])
selected = arr[[1, 3]]
selected[0] = 999
print("Selected:", selected)
print("Original:", arr)


#Boolean Indexing
#Boolean indexing also returns a copy.
arr = np.array([10, 20, 30, 40])
selected = arr[arr > 20]
selected[0] = 999
print("Selected:", selected)
print("Original:", arr)
```

**Memory Layout Basics**

NumPy stores array elements in memory. The way elements are arranged affects performance, especially for large arrays.

Two important concepts are:

- Contiguous arrays

- Memory order: C-order and F-order

```python
#A. Contiguous Arrays
#An array is contiguous when its elements are stored in a continuous, organized block of memory.
#Check memory layout using:
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(arr.flags)
```

Important flags:

Flag  ->  Meaning

C_CONTIGUOUS -> Stored in C-style row-major order

F_CONTIGUOUS -> Stored in Fortran-style column-major order

OWNDATA -> Array owns its memory

WRITEABLE -> Array can be modified

```python
#B. C-Order — Row-Major
#In C-order, elements of each row are stored together.
#For this array:
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

#The conceptual memory order is:
#1 → 2 → 3 → 4 → 5 → 6
#C-order is NumPy’s default storage order in most cases.
```

```python
#C. F-Order — Column-Major
#In Fortran-order, elements of each column are stored together.
#For the same array, the conceptual memory order is:
#1 → 4 → 2 → 5 → 3 → 6
#Create an array using Fortran order:
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
], order="F")
print(arr.flags["C_CONTIGUOUS"])
print(arr.flags["F_CONTIGUOUS"])
```

**Data Type and Memory Usage**

```python
#A. int32 vs int64
arr32 = np.array([1, 2, 3], dtype=np.int32)
arr64 = np.array([1, 2, 3], dtype=np.int64)
print(arr32.dtype)
print(arr64.dtype)
print(arr32.itemsize)  #int32 uses 4 bytes per element.
print(arr64.itemsize)  #int64 uses 8 bytes per element.
```

```python
#B. float32 vs float64
arr32 = np.array([1.5, 2.5, 3.5], dtype=np.float32)
arr64 = np.array([1.5, 2.5, 3.5], dtype=np.float64)
print(arr32.dtype)
print(arr64.dtype)
print(arr32.itemsize)
print(arr64.itemsize)
```

```python
#Using nbytes
#The .nbytes attribute gives the total memory used by the array’s elements.
arr = np.array([1, 2, 3, 4, 5], dtype=np.int32)
print(arr.nbytes)
```

```python
#Comparing Memory Usage
arr32 = np.ones(1_000_000, dtype=np.float32)
arr64 = np.ones(1_000_000, dtype=np.float64)
print(arr32.nbytes)
print(arr64.nbytes)
```

**Choosing Data Types in ML**
Use float32 When:

- Working with large numerical datasets

- Training deep learning models

- Using GPUs

- Memory efficiency is important

- float64 precision is unnecessary

Use float64 When:

- Higher numerical precision is required

- Performing scientific or financial calculations

- Small numerical errors are important
