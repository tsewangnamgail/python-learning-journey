**Advanced Reshaping**

```python
# reshape(-1, 1) -> This converts a 1D array into a column vector.
# -1 tells NumPy to automatically calculate that dimension.
import numpy as np
x = np.array([10, 20, 30, 40])
x_col = x.reshape(-1, 1)
print(x_col)
print(x_col.shape)
```

```python
#reshape(1, -1)
#This converts a 1D array into a row vector.
x = np.array([10, 20, 30, 40])
x_row = x.reshape(1, -1)
print(x_row)
print(x_row.shape)
```

**Singleton Dimensions**

A dimension whose size is 1 is called a singleton dimension.

```python
x = np.array([10, 20, 30])
print(x.shape)

#adding dimension
x = x.reshape(-1, 1)
print(x.shape)
```

**Stacking and Concatenation**

The key concept is axis.

```python
A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])

#concatenate()
#axis=0 -> combinig along row
result = np.concatenate((A, B), axis=0)
print(result)
print(result.shape)

#axis=1 -> Combine along columns.
result = np.concatenate((A, B), axis=1)
print(result)
print(result.shape)
```

**vstack() and hstack()**

```python
#vstack() -> Vertical stacking:
result = np.vstack((A, B))
print(result)
np.concatenate((A, B), axis=0)  #equivalent to above
```

```python
#hstack() -> Horizontal stacking:
result = np.hstack((A, B))
print(result)
np.concatenate((A, B), axis=1)  #equivalent to
```

**Combining Batches of Data**

```python
batch1 = np.random.default_rng(42).normal(size=(32, 3))
batch2 = np.random.default_rng(43).normal(size=(16, 3))
combined = np.concatenate((batch1, batch2), axis=0)
print(combined.shape)  #(batch_size, features)
```

**Advanced Indexing**

```python
# Integer Array Indexing -> Multiple positions can be selected simultaneously.
x = np.array([10, 20, 30, 40, 50])
indices = np.array([0, 2, 4])
print(x[indices])

# 2D Integer Indexing
A = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
rows = np.array([0, 1, 2])
cols = np.array([2, 0, 1])
print(A[rows, cols])
```

**Boolean Masks**

A Boolean mask selects elements based on a condition.

**Combining Boolean and Integer Indexing**

```python
X = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90],
    [15, 25, 35]
])
rows = X[:, 0] > 30
result = X[rows][:, [0, 2]]
print(result)
```

**Broadcasting with Higher Dimensions**

Higher-dimensional arrays are common in machine learning and deep learning.

A 3D array can represent:

(batch, height, width)

For image data with color channels, a common representation is:

(batch, height, width, channels)

For example:

(32, 224, 224, 3)

means:

32 images,224 pixels high,224 pixels wide,3 color channels

```python
#Broadcasting in 3D
X = np.zeros((2, 3, 4))
mean = np.array([1, 2, 3, 4])
result = X + mean
print(result.shape)
```

**np.einsum()**

np.einsum() stands for Einstein summation.

It provides a compact way to express operations involving:

- Multiplication
- Summation
- Dot products
- Matrix multiplication
- Tensor operations
- Axis manipulation

```python
#Vector Dot Product
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
result = np.einsum('i,i->', a, b)
print(result)
```

**Matrix Multiplication with einsum()**

```python
A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
#standard multiplication
C = A @ B
#Using einsum():
C = np.einsum('ik,kj->ij', A, B)
print(C)
```

**np.meshgrid()**

meshgrid() creates coordinate grids from coordinate arrays.

It is commonly used for:

- Mathematical functions
- Surface plots
- Contour plots
- Visualization
- Evaluating functions over a 2D grid

```python
x = np.array([1, 2, 3])
y = np.array([10, 20])
X, Y = np.meshgrid(x, y)
print(X)
print(Y)
print(X.shape)
print(Y.shape)

#Using the Grid in a Function
Z = X**2 + Y**2
print(Z)
```

**np.tile()**

tile() repeats an entire array a specified number of times.


```python
#1D array
x = np.array([1, 2, 3])
result = np.tile(x, 3)
print(result)

#For 2D arrays:
x = np.array([
    [1, 2],
    [3, 4]
])
result = np.tile(x, (2, 3))
print(result)
```

**np.repeat()**

repeat() repeats individual elements.

```python
x = np.array([1, 2, 3])
result = np.repeat(x, 3)
print(result)

#For 2D arrays, you can specify an axis:
x = np.array([
    [1, 2],
    [3, 4]
])
result = np.repeat(x, 2, axis=0)
print(result)
```

**Sliding Windows**

Sliding windows are useful when working with sequential data.

```python
#It creates overlapping windows from an array.
x = np.array([1, 2, 3, 4, 5])
windows = np.lib.stride_tricks.sliding_window_view(x, window_shape=3)
print(windows)


#Sliding Windows for Time Series
data = np.array([10, 20, 30, 40, 50, 60])
windows = np.lib.stride_tricks.sliding_window_view(data, window_shape=3)
print(windows)


#Sliding Windows with 2D Arrays
image = np.arange(16).reshape(4, 4)
print(image)
windows = np.lib.stride_tricks.sliding_window_view(image,(2, 2))
print(windows.shape)
```
